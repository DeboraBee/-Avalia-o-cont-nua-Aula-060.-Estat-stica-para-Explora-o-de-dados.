# Análise Univariada

Para os cálculos da **análise univariada** utilize a coluna `age`

## Medidas de Frequência

### Frequência absoluta

```python
# Realize o cálculo neste parágrafo
freq_abs = df['age'].value_counts().sort_index()
freq_abs
```

### Frequência relativa

```python
# Realize o cálculo neste parágrafo
freq_rel = df['age'].value_counts(normalize=True).sort_index()
freq_rel
```

### Frequência absoluta acumulada

```python
# Realize o cálculo neste parágrafo
freq_abs_acum = freq_abs.cumsum()
freq_abs_acum
```

### Frequência relativa acumulada

```python
# Realize o cálculo neste parágrafo
freq_rel_acum = freq_rel.cumsum()
freq_rel_acum
```

## Medidas de localidade

### Mediana

```python
# Realize o cálculo neste parágrafo
df['age'].median()
```

### 1º Quartil

```python
# Realize o cálculo neste parágrafo
df['age'].quantile(0.25)
```

### 3º Quartil

```python
# Realize o cálculo neste parágrafo
df['age'].quantile(0.75)
```

### Moda

```python
# Realize o cálculo neste parágrafo
df['age'].mode()
```

## Medidas de espalhamento

### Amplitude

```python
# Realize o cálculo neste parágrafo
df['age'].max() - df['age'].min()
```

### Distância entre o 3º e o 1º quartil

```python
# Realize o cálculo neste parágrafo
df['age'].quantile(0.75) - df['age'].quantile(0.25)
```

### Desvio médio absoluto

```python
# Realize o cálculo neste parágrafo
(df['age'] - df['age'].mean()).abs().mean()
```

### Variância

```python
# Realize o cálculo neste parágrafo
df['age'].var()
```

### Desvio-padrão

```python
# Realize o cálculo neste parágrafo
df['age'].std()
```

## Medidas de distribuição

### Obliquidade

```python
# Realize o cálculo neste parágrafo
df['age'].skew()
```

Explique o resultado neste parágrafo

### Curtose

```python
# Realize o cálculo neste parágrafo
df['age'].kurtosis()
```
